```python
import numpy as np
```

```python
a = np.array([[1, 2],[3, 4]])
print(a)
print(a.shape)  # (2, 2)
```

```python
r = np.random.rand(3, 4)  # uniform values in [0, 1)
r = np.round(np.random.rand(3, 4), 2)  # 2 decimal places
print(r)
print(r.shape)  # (3,4)
```

```python
ones = np.ones((3, 3))
zeros = np.zeros((3, 3))
print(ones)
print(zeros)
```

```python
m =  np.random.rand(6, 3)
m = np.round(np.random.rand(3, 4), 2) # 2 decimal places
print("Original :", m.shape)
print(m)

m_T = m.T       # or np.tranpose(m)
print("Transposed:", m_T.shape) # (3,6)
print(m_T)
```

```python
n, p, m_ = 4, 3, 5  # choose any values; the inner dimension p must match

A = np.random.rand(n, p)
B = np.random.rand(p, m_)

C = np.dot(A, B)  # or A @ B
C = np.round(np.random.rand(3, 4), 2) # 2 decimal places
print(C)
print(C.shape)  # (n, m) -> (4, 5)
```

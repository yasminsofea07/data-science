1. Find some real dataset. Load it on Colab, Kaggle etc. Try to visualize some of it

```python
df = pd.read_csv("/content/drive/MyDrive/IRIS.csv")
display(df.head())
```

2. Evaluate if the data is good or not. Does it have enough samples?



```python
print("Rows and columns:", df.shape)
print("")

print("Missing values per column:")
print(df.isnull().sum())
print("")

print("Samples per species:")
print(df["species"].value_counts())
print("")

print("Descriptive statistics:")
print(df.describe())
print("")
```

3. How many features does it have? Is it enough?

```python
print("All columns:")
print(df.columns)
print("\nNumber of columns:", len(df.columns))

# Separate inputs (X) and target (y)
print("\nInput columns (Features / X):")
print(list(df.columns.drop("species")))

print("\nTarget column (Label / y):")
print("species")
```

**Answer:** There are 4 features (input variables) in the dataset:

sepal_length

sepal_width

petal_length

petal_width

4. How many samples/observation does it have? Is it enough?

```python
print("No. of samples:", len(df))
```

5. What are the datatypes does the dataset consist of?



```python
print(df.dtypes)
```

6 and 8. How many columns have missing (null) values? If yes, which column, which rows

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())

missing_rows = df.isnull().sum(axis=1)
print("Missing values in each rows")
print(missing)
print("\nNumber of rows with any missing values:")
print(df[missing_rows > 0])  # Displays the actual data table for affected rows
print("\nNumber of rows with missing values:", (missing_rows > 0).sum())
```

7. Which column is the feature, which is the target output?

```python
# Define features (X) and target (y)
features = list(df.columns.drop('species'))
target = 'species'

print("Features (Input Columns):", features)
print("Target (Output Column):", target)
```
